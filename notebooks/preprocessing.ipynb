# Timestamp Handling

The Credit Card dataset includes a Time feature representing the number of seconds elapsed since the first recorded transaction, rather than a real-world timestamp. Similarly, the PaySim dataset uses a step variable that reflects simulated time progression.

Because these variables do not correspond to calendar-based timestamps, no conversion to datetime format was applied. Both features are retained as numerical variables and treated as contextual or ordering information rather than temporal signals.

# Outliers and Extreme Values

No automatic outlier removal was applied during data cleaning. In the Credit Card dataset, extreme values may correspond to legitimate but rare fraudulent behavior, and removing them could distort the anomaly structure of the data.

In the PaySim dataset, extreme balance values and inconsistencies are known artifacts of the simulator’s rule-based design rather than measurement errors. Removing such values at this stage could introduce unintended bias or leakage.

Therefore, outlier handling decisions are deferred to later modeling and feature selection stages, where their impact can be evaluated in context.
